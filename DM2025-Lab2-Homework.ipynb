```python
def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"http\S+|www\S+|https\S+", " ", text)
    text = re.sub(r"@[A-Za-z0-9_]+", " ", text)
    text = re.sub(r"#", " ", text)
    text = re.sub(r"\d+", " ", text)
    text = text.translate(str.maketrans("", "", string.punctuation))
    text = re.sub(r"\s+", " ", text).strip()
    return text

train_df["text_clean"] = train_df["text"].apply(clean_text)
test_df["text_clean"]  = test_df["text"].apply(clean_text)

train_df[["text", "text_clean", "emotion"]].head()
```

In this first step, I cleaned and prepared the raw text data to remove noise and transform the tweets into a more standard and usable format. The preprocessing included the following operations:

*Lowercasing:
in this one, every tweet was converted to lowercase to maintain consistency.

*Removing URLs:
the hyperlinks such as http://..., https://..., or www... were removed because they do not provide emotional information.
*Removing Mentions and Hashtags:

*Removing Numbers:
the digits were deleted since they typically do not reflect emotional content.

*Removing Punctuation:

*Removing Extra Spaces:

To convert the cleaned text into numerical features suitable for machine learning, I used the TF-IDF technique. It transforms each tweet into a vector based on the importance of its words.

Key decisions in feature engineering:

*TF-IDF Vectorizer:
is used to convert the text into numerical vectors based on frequency and importance.

*N-gram Range (1, 2):
Both unigrams and bigrams were included.

-Unigrams capture individual words (example: “happy”).

-Bigrams capture important expressions (example: “not happy”).

*Max Features : this part limited the vocabulary to the 20,000 most frequent n-grams to reduce noise and computation time.

*Stopword Removal:
common English words were removed to focus on emotional content.

Overall, we can say that the output of this step is a sparse TF-IDF matrix that encodes the importance of each word and phrase in the dataset.
